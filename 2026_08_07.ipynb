```python
img = cv2.imread('tung.webp')

cv2_imshow(img)
```

```python
img_mod = img/2
cv2_imshow(img_mod)
```

```python
img_mod_2 = img*2
cv2_imshow(img_mod_2)
```

```python
B, G, R = cv2.split(img)
cv2_imshow(B)

# passar só uma camada faz imprimir em grayscale
```

```python
plt.subplot(1,3,1)
plt.imshow(B, cmap='gray')
plt.axis('off') # tira as linhas e números do plot
plt.title('Azul (B)')

plt.subplot(1,3,2)
plt.imshow(G, cmap='gray')
plt.axis('off')
plt.title('Verde (G)')

plt.subplot(1,3,3)
plt.imshow(R, cmap='gray')
plt.axis('off')
plt.title('Vermelho (R)')
```

```python
n = np.arange(0,20)

plt.subplot(2,2,1)
plt.plot(n,np.sin(n))

plt.subplot(2,2,2)
plt.plot(n,np.cos(n))

plt.subplot(2,2,3)
plt.plot(n)

plt.subplot(2,2,4)
plt.plot(n*n)
```
