```python
img_bgr = cv2.imread('luladroid.jpg')
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
plt.imshow(img_gray, cmap='gray')
plt.show()
```

```python
altura = img_gray.shape[0]
largura = img_gray.shape[1]
print(altura, largura)
```

```python
dx = 50
dy = 30
translada = np.zeros((altura, largura), dtype= np.uint8)

for x in range (largura):
  for y in range (altura):

    novo_x = x + dx
    novo_y = y + dy

    if novo_x < largura and novo_y < altura:
      translada[novo_y,novo_x] = img_gray[y,x]
plt.figure(figsize=(7,5))
plt.imshow(translada, cmap='gray')
plt.show()
```

```python
wraparound = np.zeros((altura, largura), dtype= np.uint8)

for x in range (largura):
  for y in range (altura):

    novo_x = (x + dx)% largura
    novo_y = (y + dy) % altura
    wraparound[novo_y,novo_x] = img_gray[y,x]

plt.figure(figsize=(7,5))
plt.imshow(wraparound, cmap='gray')
plt.show()
```

```python
sx = 2
sy = 0.5

nova_largura = int(largura * sx)
nova_altura = int(altura * sy)

print(nova_altura,nova_largura)

```

```python
escalada = np.zeros((nova_altura, nova_largura), dtype= np.uint8)

for y in range(nova_altura):
    for x in range(nova_largura):

        novo_x = min(int(x / sx), img_gray.shape[1] - 1)
        novo_y = min(int(y / sy), img_gray.shape[0] - 1)

        escalada[y, x] = img_gray[novo_y, novo_x]

plt.figure(figsize=(7,5))
plt.imshow(escalada, cmap='gray')
plt.show()
```

```python
espelhada = np.zeros((altura, largura), dtype= np.uint8)

for x in range (largura):
  for y in range (altura):

    novo_x = largura - x - 1
    novo_y = y

    espelhada[y,x] = img_gray[novo_y,novo_x]

plt.figure(figsize=(7,5))
plt.imshow(espelhada, cmap='gray')
plt.show()
#
```
